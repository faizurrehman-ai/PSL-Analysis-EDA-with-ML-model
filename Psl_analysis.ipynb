```python
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report
from xgboost import XGBClassifier
import matplotlib.pyplot as plt
import seaborn as sns

# Data Aggregation
inn1 = data[data['inning'] == 1].groupby('match_id').agg({
    'total_runs': 'sum',
    'is_wicket': 'sum',
    'over': 'max',
    'batting_team': 'first',
    'bowling_team': 'first',
    'venue': 'first',
    'season': 'first',
    'winner': 'first'
}).reset_index()

inn1.columns = ['match_id', 'runs', 'wickets', 'overs', 'team1', 'team2', 'venue', 'season', 'winner']

# Run rate (important)
inn1['run_rate'] = inn1['runs'] / inn1['overs']

# Pressure (wickets impact)
inn1['pressure'] = inn1['wickets'] / 10

# Performance score
inn1['performance'] = inn1['runs'] - (inn1['wickets'] * 8)


all_teams = sorted(list(set(inn1['team1']) | set(inn1['team2'])))
team_stats = {team: {'wins': 0, 'games': 0} for team in all_teams}

t1_wr, t2_wr = [], []

for _, row in inn1.iterrows():
    wr1 = team_stats[row['team1']]['wins'] / team_stats[row['team1']]['games'] if team_stats[row['team1']]['games'] > 0 else 0.5
    wr2 = team_stats[row['team2']]['wins'] / team_stats[row['team2']]['games'] if team_stats[row['team2']]['games'] > 0 else 0.5

    t1_wr.append(wr1)
    t2_wr.append(wr2)

    team_stats[row['team1']]['games'] += 1
    team_stats[row['team2']]['games'] += 1

    if row['winner'] == row['team1']:
        team_stats[row['team1']]['wins'] += 1
    else:
        team_stats[row['team2']]['wins'] += 1

inn1['t1_strength'] = t1_wr
inn1['t2_strength'] = t2_wr

# Target
inn1['target'] = (inn1['winner'] == inn1['team1']).astype(int)

# Encoding 
le = LabelEncoder()
inn1['v_n'] = le.fit_transform(inn1['venue'])
inn1['t1_n'] = le.fit_transform(inn1['team1'])
inn1['t2_n'] = le.fit_transform(inn1['team2'])

# Features
X = inn1[[
    'runs', 'wickets', 'run_rate', 'pressure', 'performance',
    'v_n', 't1_n', 't2_n', 't1_strength', 't2_strength'
]]

y = inn1['target']

# Split 
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=False
)

# Model (Boosted) 
model = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.9,
    colsample_bytree=0.9,
    random_state=42,
    eval_metric='logloss'
)

print("Training Improved Model...")
model.fit(X_train, y_train)

# Result's 
y_pred = model.predict(X_test)
acc = (y_pred == y_test).mean() * 100

print(f"\nFinal Accuracy: {acc:.2f}%")

# Visualization
plt.figure(figsize=(10, 4))

# Confusion Matrix
plt.subplot(1, 2, 1)
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Greens')
plt.title('Confusion Matrix')
plt.xlabel('Predicted'); plt.ylabel('Actual')

# Feature Importance
plt.subplot(1, 2, 2)
importance = model.feature_importances_
plt.barh(X.columns, importance)
plt.title('Feature Importance')

plt.tight_layout()
plt.show()
```
